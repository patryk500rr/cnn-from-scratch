```python
from layers import MaxPool2d, Conv2d, Dense, Flatten, ReLU, Dense, GlobalAvgPool, MaxPool2d_naive, AvgPool2d, AvgPool2d_naive
from model import Sequential
import numpy as np
```

```python
rng = np.random.default_rng(123)


model_maxpool_naive = Sequential([
    Conv2d(8, f=3, s=1, p='same'), 
    ReLU(), 
    AvgPool2d_naive(), 
    Conv2d(16, f=3, s=1, p='same'), 
    ReLU(), 
    AvgPool2d_naive(), 
    Flatten(), 
    Dense(64),
    ReLU(),
    Dense(10)], 
    input_shape=(28, 28, 3), rng=rng)

model_maxpool_naive.build()
```

```python
_, shapes_naive = model_maxpool_naive.summary()
```

```python
model_maxpool = Sequential([
    Conv2d(8, f=3, s=1, p='same'), 
    ReLU(), 
    AvgPool2d(), 
    Conv2d(16, f=3, s=1, p='same'), 
    ReLU(), 
    AvgPool2d(), 
    Flatten(), 
    Dense(64),
    ReLU(),
    Dense(10)], 
    input_shape=(28, 28, 3), rng=rng)

model_maxpool.build()
```

```python
_, shapes = model_maxpool.summary()
```

```python
shapes_naive.values() == shapes.values()
```

```python
X = rng.standard_normal((1, 28, 28, 3))

my_maxpool = MaxPool2d(f=3, s=1)
my_maxpool.build((28, 28, 3))

my_maxpool_naive = MaxPool2d_naive(f=3, s=1)
my_maxpool_naive.build((28, 28, 3))

Z = my_maxpool.forward(X)
Z_naive = my_maxpool_naive.forward(X)
np.allclose(Z, Z_naive)
```

```python
dZ = rng.standard_normal(Z.shape)

dX = my_maxpool.backward(dZ)
dX_naive = my_maxpool_naive.backward(dZ)

np.allclose(dX, dX_naive)
```
